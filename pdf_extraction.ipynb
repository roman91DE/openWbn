```python
from pathlib import Path
import requests
import pandas as pd
import tabula
```

```python
%pip install tabula-py
```

```python
url = """https://www.wiesbaden.de/medien/downloads/leben-in-wiesbaden/amt-12/020-Statistisches-Jahrbuch-Gesamtausgabe-2024.pdf"""
```

```python
cw_dir = Path().cwd()
tmp_dir = cw_dir / "tmp"
tmp_dir.mkdir(exist_ok=True, parents=True)
```

```python
%ls
```

```python
resp = requests.get(url=url)
```

```python
print(
    "http status code: ",
    resp.status_code
)
```

```python
resp.text[:100]
```

```python
f = tmp_dir / "jahrbuch.pdf"
f.write_bytes(resp.content)
```

```python
%ls ./tmp/
```

```python
!du -h ./tmp/jahrbuch.pdf
```

```python
tables = tabula.read_pdf(f, pages="all")
```

```python
len(tables)
```

```python
tables.sort(key=len, reverse=True)
```
